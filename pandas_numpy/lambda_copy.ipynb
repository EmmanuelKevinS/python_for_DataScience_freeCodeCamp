### Lambda function

```python
# basic function
def sum_values(a,b):
    c = a + b
    return c
```

```python
print(sum_values(3,4))
```

```python
#Lambda function (one liner)
sum_values_lambda = lambda d,e: d+e
```

```python
sum_values_lambda(3,4)
```

### Apply + Lambda function

```python
#Use lambda function to convert "height_cm" to metres
lambda_height = df['height_cm'].apply(lambda m:(m/100))
lambda_height.head()
```

```python
#Use lambda function to convert "long_name" to upper case
lambda_upper_case = df['long_name'].apply(lambda ln:ln.upper())
lambda_upper_case.head()
#alternative with str attribute
non_lambda_upper = df['long_name'].str.upper()
non_lambda_upper.head()
```

```python
df.dtypes
```

```python
df['dob'] = df['dob'].astype('datetime64')
#use lambda function to get the year of the "dob" series
year_lambda = df['dob'].apply(lambda x:x.year)
year_lambda.head()
```

```python
df['height_cm'] = df['height_cm'].apply(lambda x:x/100)
```

```python
df.rename(columns={'height_cm': 'height_m'}, inplace=True)
# apply lambda function to calculateBMI(weight/height*height)
lambda_BMI = df.apply(lambda x:x['weight_kg']/(x['height_m']**2),axis=1)
lambda_BMI
```

### Making a copy with copy()

#### Deep = True -> changes to original DF will NOT be reflected in the copy

```python
df_copy = df.copy()
df_copy.head()
```

#### Deep = False -> changes to original DF will be reflected in the copy

```python
df_shallow_copy = df_copy.copy(deep=False)
df_copy.loc['L. Messi','height_m'] = 1.80
```
