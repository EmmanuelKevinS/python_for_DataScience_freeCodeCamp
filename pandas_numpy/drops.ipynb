### Create random sample with sample method

```python
df.head()
```

### sample()

```python
#Extract 10 rando elements form nationality column
df['nationality'].sample(10, random_state=76)
```

```python
#extract a random 20% sample of the dataframe
#// -> floor division because 18278 is not divisible by 5
df.sample(18278//5).sort_values(by='overall',ascending=False)
```

```python
#same as previous by frac method
df.sample(frac=0.2,random_state=35).sort_values(by='overall',ascending=False)
```

```python
#upsample: increase the samping rate ->frac >1(Note: replace parameter has to be true for frac parameter > 1)
df.sample(frac=2,replace=True,random_state=23).sort_values(by=['overall'],ascending=False)
```

### Filtering a df with query() method

### Query()

```python
# select players older than 34
df.query('age>34')
# write the equivalent boolean slicing
df[df['age'] > 34]
```

```python
# select players older than 34 from italy
df.query('age > 34 and nationality == "Italy" ')
# write equivalent booleean slicing
ageroni = df['age'] > 34
italianoooo = df['nationality'] == 'Italy'
df[ageroni & italianoooo]
```

```python
# add not operator to the first example ~ !=
df.query('not(age>34)')

# write equivalent boolean slicing
df[~(df['age'] > 34)]
```

```python
# convert height to metres
df.rename(columns={'height_cm': 'height_m'}, inplace=True)
df['height_m'] = df['height_m'] / 100
df.head()
```

```python
# select all with height > 1.8m
df.query('height_m>1.8')

# write equivalent boolean slicing
df[df['height_m'] > 1.8]
```

```python
# select players born after 1990
# check data type
df.dtypes
# convert dob column to datetime type
df['dob'] = pd.to_datetime(df['dob'], errors='coerce')
df.dtypes
# query
df['dob'].dt.year
df.query('dob.dt.year>1990')
# write equivalent boolean slicing
df[df['dob'].dt.year > 1990] 

```
