## Selecting data

```python
import pandas as pd
import numpy as np
df = pd.read_csv('DataExtraction/players_20.csv',encoding='utf-8')
df.head()
```

```python
#set index
df = df.set_index('short_name')
df.head()
```

```python
#select columns
df = df[['long_name','age','dob','height_cm','weight_kg','nationality','club','overall']]
df.head()
```

## Selecting with a single value

loc[row_label,column_label]

```python
#get all data about L.Messi
df.loc['L. Messi']
```

```python
#get the height of L.Messi
df.loc['L. Messi','height_cm']
```

```python
#Weight of C.Ronaldo
df.loc['Cristiano Ronaldo','weight_kg']
```

```python
#get all rows in height column
df.loc[:,'height_cm']
```

```python
#Get all columns that correspond to the name L. Messi
df.loc['L. Messi',:]
```

## Selecting with a list of values

```python
#Get all data about L. Messi and Cristiano Ronaldo
df.loc[['L. Messi','Cristiano Ronaldo']]
```

```python
#get both their heights
df.loc[['L. Messi','Cristiano Ronaldo'],'height_cm']
```

```python
#height and weight of L. Messi
df.loc['L. Messi',['height_cm','weight_kg']]
```

```python
#height and weight of both
df.loc[['L. Messi','Cristiano Ronaldo'],['height_cm','weight_kg']]
```

## Selecting a range of data with a slice
start:stop:step ("Stop" INDEX IS INCLUDED)

```python
#Slice column labels
players = ['L. Messi','Cristiano Ronaldo']
df.loc[players,'age':'club']
```

```python
df.head(15)
```

```python
#Slice index labels
columns = ['age','dob','height_cm','weight_kg']

df.loc['Neymar Jr':'Alisson',columns]
```

## Selecting with conditions

```python
#Selecting players with height > 199cm
columns = ['age','dob','height_cm','weight_kg','overall']
df.loc[df['height_cm']>199,columns].sort_values(by='overall',ascending=False)
```

```python
#Selecting players with height > 199cm from England
df.loc[(df['height_cm']>199) & (df['nationality'] == 'England'),columns].sort_values(by='overall',ascending=False)
```

```python
#Selecting players with height > 199cm from England OR Sweden
df.loc[(df['height_cm']>199) & ((df['nationality'] == 'England') | (df['nationality'] == 'Sweden')),columns].sort_values(by='overall',ascending=False)
```

## iloc method

### Selecting with a single value
iloc[row_position,column_position]

```python
#get height of L. Messi
row_num = df.index.get_loc('L. Messi')
column_num = df.columns.get_loc('height_cm')

df.iloc[row_num,column_num] #same as df.iloc[0,3]
```

```python
#weight of Cristiano
df.iloc[df.index.get_loc('Cristiano Ronaldo'),df.columns.get_loc('weight_kg')]
```

```python
#get all rows in the 'height_cm' column
df.iloc[:,3]
```

```python
#get all messi columns
df.iloc[row_num,:]
```

### Selecting with a list of values


```python
#Get all data about Messi and CR7
df.iloc[[0,1],:] 
```

```python
#Get height of messi and ronaldo
df.iloc[[0,1],3] 
```

```python
#height and weight of messi and ronaldo
df.iloc[[0,1],[3,4]]
```

### Selecting range of data with a slice
start:stop (Stop is excluded)

```python
# slice columns labels age/club
# players = ['L. Messi','Cristiano Ronaldo']
players = [0,1]
df.iloc[players, 1:7]
```

```python
#slice index labels(top 1 to 10)
#columns = ['age','dob','height_cm','weight_kg']
# get top 1 to 10 players
df.iloc[0:10,1:5]
```

### selecting with conditions
iloc cannot accept boolean series but only a boolean list. have to use list() function to convert series into a boolean list

```python
#one condition: select players with height >199cm
#columns = ['age','dob','height_cm','weight_kg'] -> 1:5
list_from_series = list(df['height_cm']>199)
df.iloc[list_from_series,1:5]
```

```python
#multiple conditions
#players above 199cm from england and sweden
conditions_list = list((df['height_cm']>199) & ((df['nationality'] == 'England') | (df['nationality'] == 'Sweden')))
df.iloc[conditions_list,1:5]                
```
