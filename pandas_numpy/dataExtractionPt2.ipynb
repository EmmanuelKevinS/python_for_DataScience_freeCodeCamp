```python
import pandas as pd
import numpy as np
df = pd.read_csv('DataExtraction/players_20.csv',encoding='utf-8')
df.head()
```

```python
df = df.set_index('short_name')
df = df[['long_name','age','dob','height_cm','weight_kg','nationality','club','overall']]
df.head()
```

### Set value to one cell

```python
#Update L. Messi height
df.loc['L. Messi', 'height_cm'] = 175
df.head(3)

```

### Set value to entire column

```python
df.loc['L. Messi','height_cm'] = 170
early_backup = df.copy()
early_backup
```

```python
#set all players height to 193
df.loc[:,'height_cm'] = 193
df.head()
```

### Set value to entire row

```python
early_backup.head(3)
```

```python
df = early_backup
```

### Set value to multiple cells

```python
#set values to all items matching the list of labels
df.loc[['L. Messi','Cristiano Ronaldo'],['height_cm']] = 172
df.head()
```

```python
df.loc['L. Messi','height_cm'] = 170
df.loc[['Cristiano Ronaldo'],['height_cm']] = 187
df.head()
```

### Set values for rows matching a condition

```python
columns = ['age','dob','height_cm','weight_kg']

df['height_cm'] > 180

df.loc[df['height_cm'] > 180, columns] = 0

df
```
