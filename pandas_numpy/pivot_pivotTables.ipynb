### pivot_table()

```python
df_sales = pd.read_excel('PivotTable/supermarket_sales.xlsx')
```

```python
df_sales.head()
```

```python
#make a pivot table and add an aggregate function 
df_sales.pivot_table(index='Gender',aggfunc='mean') #only columns with numerical data are displayed
```

```python
#make a pivot table, add an aggregate function and select some columns
df_sales.pivot_table(index='Gender',values=['Quantity','Total'],columns='Product line',aggfunc='sum').round(1)
```
